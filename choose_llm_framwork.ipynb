### Choosing LLM framework

```python
from openai import OpenAI
from IPython.display import display, Markdown
```

```python
import sys
print(sys.executable)
```

```python
openai = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
messages = [
    {"role": "system" , "content":"You are a funny chatbot"},
    {"role":"user", "content": "Tell me about AI revolution in coming years"}
]
```

```python
response = openai.chat.completions.create(model = "llama3.1:latest", messages=messages)
clear_output = response.choices[0].message.content
display(Markdown(clear_output))
```

```python
### Langchain

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model = "gemma:2b",base_url ="http://localhost:11434/v1", api_key = "ollama")
response = llm.invoke("Tell me about AI revolution in coming years")
display(Markdown(response.content))
```

```python
from litellm import completion
from IPython.display import Markdown, display

response = completion(
    model="ollama/gemma:2b",
    messages=[{"role": "user", "content": "Tell me about AI revolution in coming years"}]
)

reply = response['choices'][0]['message']['content']

display(Markdown(reply))
```

#### *LiteLLM provides a unified interface to interact with multiple LLM providers, while LangChain helps build end-to-end LLM applications like chatbots and RAG systems by managing memory, tools, and workflow*
