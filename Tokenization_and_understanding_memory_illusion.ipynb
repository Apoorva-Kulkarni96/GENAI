### Tokenizing with code

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")
tokens = encoding.encode("Hi, my name is Apoorva")

print(tokens)
print(encoding.decode(tokens))
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")
```

```python
from openai import OpenAI
openai = OpenAI(base_url="http://localhost:11434/v1",api_key="ollama")
```

```python
messages = [
    {"role" : "system", "content" : "You are a helpful assistant"},
    {"role" : "user", "content": "Hi! I am Apoorva"},
    {"role" : "assistant", "content" : "Hi Apoorva, How can I help you?"},
    {"role" : "user","content":"What is my name?"}
    ]
```

```python
response = openai.chat.completions.create(model="llama3.1:latest", messages=messages)
response.choices[0].message.content
```

### To Summarize

1. Every call to an LLM is stateless
2. We pass in the entire converstion so far in the input prompt, every time
3. This gives an illusion that LLM has memory - it apparently keeps context of conversation
4. But this is a trick; its a byproduct of providing the entire converstions, every time 
5. An LLM just predicts the most likely the next token in sequence; if sequence contain "My name is Apoorva" and later "What is my name?" then it will predict Apoorva 

The ChatGPT product uses the exactly this trick - everytime you send a message, its entire conversation gets passed in

Note : We  have to pay extra each time for all conversation so far

### Understanding the Context Window
A model’s context window consists of three components:

* Input tokens – the tokens in your current prompt.
* Conversation history – previous messages included to maintain context.
* Output tokens – the tokens generated in the response.

You can reduce token usage by trimming or removing conversation history, but this may affect response quality because the model has less context to work with.

Additionally, if you are using a **reasoning model**, the tokens consumed during its internal reasoning or thought process are also counted toward the total token usage. This can significantly increase overall cost and token consumption.

Caching : if you send same input twice within few minutes, then cost (atleast for gpt) will be less
