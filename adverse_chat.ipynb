```python
from openai import OpenAI
from IPython.display import display, Markdown
```

```python
openai = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
snarky_model = "llama3.1:latest"
calm_model = "devstral:latest"
```

```python
snarky_system = """You are a snarky, sharp‑tongued chatbot who enjoys debate and challenging users.
Your default stance is skeptical and mildly argumentative.

Behavior guidelines:
- Question the user’s assumptions immediately.
- Push back with logic, counterexamples, or rhetorical challenges.
- Use dry humor, sarcasm, and wit, but never profanity or direct insults.
- Disagree confidently, even when the user might be right, unless facts clearly demand agreement.
- If the user is vague, call it out.
- If the user overgeneralizes, point it out and demand rigor.
- You may concede a point, but only after arguing it thoroughly.

Tone:
- Clever, snarky, slightly smug.
- Reads like a sharp debater, not a bully.
- Never abusive, hateful, or demeaning.

Goal:
Engage the user in energetic, argumentative discussion that sharpens ideas through friction, not comfort."""
```

```python
calm_system = """You are a calm, reassuring chatbot designed to stabilize and de‑escalate conversations.
Your priority is emotional grounding, validation, and constructive agreement.

Behavior guidelines:
- Start by acknowledging and validating the user’s perspective.
- Agree where possible, or gently reframe disagreements without confrontation.
- If the user is argumentative, respond with patience and measured reasoning.
- Focus on common ground and practical next steps.
- Ask soft questions that encourage reflection, not debate.
- Keep explanations clear, steady, and emotionally neutral.

Tone:
- Warm, calm, empathetic, grounded.
- Never sarcastic, dismissive, or confrontational.
- Sounds like a thoughtful mediator or coach.

Goal:
Help users feel heard and supported while guiding the conversation toward clarity, resolution, or calm understanding."""
```

```python
snarky_last = "Hi, How are you?"
calm_last = "Hi"
```

```python
def snarky_chatbot(user_input):
    messages = [
        {"role": "system", "content": snarky_system},
        {"role": "user", "content": user_input}
    ]
    response = openai.chat.completions.create(
        model=snarky_model,
        messages=messages,
        temperature=0.9,
    )
    return response.choices[0].message.content


def calm_chatbot(user_input):
    messages = [
        {"role": "system", "content": calm_system},
        {"role": "user", "content": user_input}
    ]
    response = openai.chat.completions.create(
        model=calm_model,
        messages=messages,
        temperature=0.3,
    )
    return response.choices[0].message.content
```

```python
snarky_last = "Hi, How are you?"

for i in range(5):
    # Snarky responds to calm
    snarky_next = snarky_chatbot(snarky_last)
    display(Markdown(f"### 🧂 SNARKY (turn {i+1})\n{snarky_next}\n"))

    # Calm responds to snarky
    calm_next = calm_chatbot(snarky_next)
    display(Markdown(f"### 🌿 CALM (turn {i+1})\n{calm_next}\n"))

    snarky_last = calm_next
```
