### Write a MLP from Scratch 


```python
import numpy as np

class MLP:
    def __init__ (self, input_dim, hidden_dim, output_dim, epoch=100, learning_rate=0.01):
        self.W1 = np.random.randn(input_dim, hidden_dim)
        self.b1 = np.zeros((1, hidden_dim))
        self.W2 = np.random.randn(hidden_dim, output_dim)
        self.b2 = np.zeros((1, output_dim))
        
        self.epoch = epoch
        self.learning_rate = learning_rate

    def relu(self, x):
        return np.maximum(0, x)
    
    def relu_derivative(self, x):
        return np.where(x > 0, 1, 0)
    
    def sigmoid(self, x):
        return 1/(1+np.exp(-x))
    
    def sigmoid_derivative(self, x):
        sx = self.sigmoid(x)
        return sx * (1 - sx)

    def fit(self, x, y):
        for _ in range(self.epoch):
            # Forward pass
            z1 = np.dot(x, self.W1) + self.b1
            a1 = self.relu(z1)
            z2 = np.dot(a1, self.W2) + self.b2
            a2 = self.sigmoid(z2)

            # Backpropagation
            error = a2 - y
            d_z2 = error * self.sigmoid_derivative(z2)
            dW2 = np.dot(a1.T, d_z2)
            db2 = np.sum(d_z2, axis=0, keepdims=True)

            d_z1 = np.dot(d_z2, self.W2.T) * self.relu_derivative(z1)
            dW1 = np.dot(x.T, d_z1)
            db1 = np.sum(d_z1, axis=0, keepdims=True)

            # Update weights
            self.W1 -= self.learning_rate * dW1
            self.b1 -= self.learning_rate * db1
            self.W2 -= self.learning_rate * dW2
            self.b2 -= self.learning_rate * db2

    def predict(self, x):
        z1 = np.dot(x, self.W1) + self.b1
        a1 = self.relu(z1)
        z2 = np.dot(a1, self.W2) + self.b2
        return self.sigmoid(z2)
```
