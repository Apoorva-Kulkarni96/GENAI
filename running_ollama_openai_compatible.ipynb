### Running Ollama Locally with OpenAI Compatibile Endpoints

```python
import requests

requests.get("http://localhost:11434").content
```

```python
from openai import OpenAI
```

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url = OLLAMA_BASE_URL, api_key = "ollama")
```

```python
request = ollama.chat.completions.create(model="llama3.1:latest", messages=[{"role": "user", "content": "Tell me a Fun Fact"}])
```

```python
print(request.choices[0].message.content)
```
