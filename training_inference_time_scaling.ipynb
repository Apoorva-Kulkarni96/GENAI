### Training vs. Inference Time Scaling
There are two main ways to improve a model’s performance on difficult reasoning tasks:
1. Scale at Inference Time (Bigger Reasoning Budget)

A small model with limited reasoning depth may get the answer wrong.
But if you increase its reasoning steps (e.g., chain‑of‑thought, tree search, or test‑time compute), the same model can solve it correctly.
This shows how test‑time compute scaling improves performance without retraining.

2. Scale at Training Time (Bigger Model or More Training)

If you increase model size or train a larger version of the model, the answers improve even without additional reasoning depth.
This shows how model scale and training compute also improve reasoning quality.

```python
from openai import OpenAI
openai = OpenAI(base_url="http://localhost:11434/v1",api_key="ollama")
```

```python
easy_puzzle =[
    {"role":"user", "content":"You toss 2 coins. One of them is heads. What's the probability the other is tail? Answer ONLY in a probability format like 1/3, 2/3, 0.25 — no explanation."}
]
```

```python
response = openai.chat.completions.create(model="deepseek-r1:8b", messages=easy_puzzle)
response.choices[0].message.content
```

```python
response = openai.chat.completions.create(model="deepseek-r1:8b", messages=easy_puzzle, reasoning_effort="low")
response.choices[0].message.content
```
