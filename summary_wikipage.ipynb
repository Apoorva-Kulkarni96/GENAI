```python
from openai import OpenAI
import requests

```

```python
system_prompt = """
You are a snarky assistant that analyzes the content of website,
and provide short, snarky, homorous summary, ignorging text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with markdown.
"""

user_prompt = """
Here are the contents of the website.
Provide a short summary of this website.
If it includes news and announcements, then summarize these too.
"""
```

```python
def message_for(website):
    return [{"role":"sytem", "content":system_prompt},{"role":"user", "content":user_prompt+website}]
```

```python
def fetch_wiki_plaintext(WIKI_API):
    headers = {
        "User-Agent": "Apoorva-AI-Project/1.0 (learning project)"
    }

    r = requests.get(WIKI_API, headers=headers, timeout=20)
    r.raise_for_status()

    data = r.json()
    pages = data.get("query", {}).get("pages", {})

    for page in pages.values():
        return page.get("extract", "")
```

```python
def summarize(OLLAMA_URL, MODEL, WIKI_API):
    ollama = OpenAI(base_url=OLLAMA_URL, api_key="ollama")
    text = fetch_wiki_plaintext(WIKI_API)[:5000]
    response = ollama.chat.completions.create(model= MODEL, messages=message_for(text))
    return response.choices[0].message.content

```

```python
def main():
    WIKI_TITLE = "Artificial_intelligence" 
    WIKI_API = f"https://en.wikipedia.org/w/api.php?action=query&prop=extracts&explaintext&format=json&titles={WIKI_TITLE}"
    OLLAMA_URL = "http://localhost:11434/v1"
    MODEL = "gemma:2b"
    summary = summarize(OLLAMA_URL,MODEL,WIKI_API)
    print(summary)
    
```

```python
if __name__ == "__main__":
    main()
```
