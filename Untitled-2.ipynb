```python
import requests

WIKI_TITLE = "Artificial_intelligence"  # derived from your URL
WIKI_API = f"https://en.wikipedia.org/w/api.php?action=query&prop=extracts&explaintext&format=json&titles={WIKI_TITLE}"
OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL = "llama3"   # change to a local model you prefer

def fetch_wiki_plaintext():
    headers = {
        "User-Agent": "Apoorva-AI-Project/1.0 (learning project)"
    }

    r = requests.get(WIKI_API, headers=headers, timeout=20)
    r.raise_for_status()

    data = r.json()
    pages = data.get("query", {}).get("pages", {})

    for page in pages.values():
        return page.get("extract", "")

def ollama_generate(prompt: str) -> str:
    payload = {"model": MODEL, "prompt": prompt, "stream": False}
    r = requests.post(OLLAMA_URL, json=payload, timeout=180)
    r.raise_for_status()
    return r.json().get("response", "")

def chunk_text(s: str, max_chars: int = 8000):
    # Conservative chunk size for typical 8B–13B models in common quantizations.
    # Adjust upward/downward based on your model / context length.
    return [s[i:i+max_chars] for i in range(0, len(s), max_chars)]

def summarize_chunk(i: int, total: int, text: str) -> str:
    prompt = (
        f"You are summarizing chunk {i}/{total} of a Wikipedia article.\n"
        f"Provide 5–8 factual bullet points capturing key information in this chunk only.\n"
        "Avoid redundancy; do not invent details.\n\n"
        f"{text}"
    )
    return ollama_generate(prompt)

def merge_summaries(summaries: list) -> str:
    joined = "\n\n".join(
        [f"Chunk {i+1} summary:\n{sm}" for i, sm in enumerate(summaries)]
    )
    prompt = (
        "You are given partial summaries of a Wikipedia article. Merge them into:\n"
        "- 6–8 consolidated bullet points (unique, non-overlapping)\n"
        "- One TL;DR line at the end\n"
        "Preserve only the most important facts and keep it concise.\n\n"
        f"{joined}"
    )
    return ollama_generate(prompt)

def main():
    full_text = fetch_wiki_plaintext()

    # Quick exit for small pages
    if len(full_text) <= 8000:
        final = ollama_generate(
            "Summarize the following Wikipedia article into 6–8 bullets and a TL;DR:\n\n" + full_text
        )
        print(final)
        return

    chunks = chunk_text(full_text, max_chars=8000)
    partials = []
    for i, ch in enumerate(chunks, start=1):
        partial = summarize_chunk(i, len(chunks), ch)
        partials.append(partial)

    final_summary = merge_summaries(partials)
    print(final_summary)

if __name__ == "__main__":
    main()
```
