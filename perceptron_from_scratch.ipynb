### Write a Preceptron using numpy 

```python
import numpy as np

class Perceptron:
    def __init__(self, input_dim, learning_rate=0.01, epochs=100):
        self.W = np.zeros(input_dim)
        self.b = 0
        self.lr = learning_rate
        self.epochs =epochs

    #activation fucntion    
    def step(self, x):
        return 1 if (x >= 0) else 0
    
    def fit(self, X, y):
        for _ in range(self.epochs):
            for i in range(X.shape[0]):
                linear_output = np.dot(X[i],self.W) + self.b
                y_pred = self.step(linear_output)

                error = y[i] - y_pred

                self.W += self.lr * error * X[i]
                self.b += self.lr * error


    def predict(self, X):
        linear_output = np.dot(X, self.W) +self.b
        return np.where(linear_output >=0, 1,0)
```

### Example Test Code (AND Gate)





```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])
X,y
```

```python
model = Perceptron(input_dim=2, learning_rate=0.1, epochs=10)
model.fit(X,y)
```

```python
prediction = model.predict(X)
print("Predictions:", prediction)
print("Actual:", y)
```

```python
### Accuracy

accuracy = np.mean(prediction == y)
print("Accuracy:", accuracy)
```
